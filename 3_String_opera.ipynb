## Group3. String Operations

### Q1. Find all customers whose name starts with 'A'

```python
df[df['Customer Name'].str.startswith('A')]
```

### Q2. Find all products that contain the word 'Chair'

```python
df[df['Product Name'].str.contains('Chair')]
```

### Q3. Find all orders where Sub-Category contains 'Book' (case insensitive)

```python
df[df['Sub-Category'].str.contains('book', case = False)]
```

### Q4. Count how many unique customers exist in the dataset

```python
df['Customer Name'].nunique()
```

### Q5. Show the top 5 most frequent cities by order count

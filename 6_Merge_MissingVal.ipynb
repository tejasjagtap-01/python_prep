### Group By the df

```python
# df1 - order summary

df1 = df.groupby('Order ID').agg(
    total_sales = ('Sales','sum'),
    total_profit = ('Profit', 'sum')
).reset_index()
```

```python
df1
```

```python
df1.head()
```

```python
# df2 - customer info per order

df2 = df[['Order ID', 'Customer Name', 'Segment', 'Region']].drop_duplicates()
```

```python
df2
```

---

### Q1. Merge df1 and df2 on Order ID (inner join)

```python
merge = pd.merge(df1,df2, on='Order ID', how='inner')
merge
```

```python
merge.shape
```

### Q2. Check for missing values in the merged result

```python
merge.isnull().sum()
```

### Q3. If any missing — fill with appropriate values

```python
merge['Customer Name'].fillna('Unknown', inplace=True)
merge['total_sales'].fillna(0,inplace=True)
```

```python
merge['Customer Name'] = merge['Customer Name'].fillna('Unknown')
merge['total_sales'] = merge['total_sales'].fillna(0)
```

### Q4. How many orders exist in df1 but NOT in df2? (left anti join)

```python
left_join = pd.merge(df1,df2, on='Order ID', how='left')
left_join
```

```python
not_in_df2 = left_join[left_join['Customer Name'].isnull()]
```

```python
not_in_df2
```

```python
print(f"Orders only in df1: {len(not_in_df2)}")
```

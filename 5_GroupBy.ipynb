### Q1. Total sales by Region

```python
df.groupby('Region')['Sales'].sum()
```

### Q2. Average profit by Category

```python
df.groupby('Category')['Profit'].mean()
```

### Q3. Both total sales AND average profit by Category in one query

```python
df.groupby('Category').agg(
    total_sales = ('Sales','sum'),
    average_profit = ('Profit', 'mean')
).round(2)
```

```python
df.groupby('Category').agg(
    total_sales = ('Sales','sum'),
    average_profit = ('Profit', 'mean')
)
```

### Q4. Top 3 Sub-Categories by total sales

```python
df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending= False) .head(3).round(1)
```

### Q5. Number of orders and total sales per Customer Segment

```python
df.groupby('Segment').agg(
    order_count = ('Order ID', 'count'),
    total_sales = ('Sales','sum')
)
```

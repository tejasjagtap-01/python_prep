### Q1. Convert Order Date to datetime first

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df.dtypes
```

### Q2. Extract year and month from Order Date

```python
df['Order Year'] = df['Order Date'].dt.year
```

```python
df.head(1)
```

```python
df['Order Month'] = df['Order Date'].dt.month
```

```python
df.head(1)
```

### Q3. Create a pivot table showing total sales by Category (rows) and Region (columns)

```python
pd.pivot_table(
    df,
    values='Sales',
    index='Category', 
    columns='Region',
    aggfunc = 'sum'
)
```

```python
pd.pivot_table(
    df,
    values='Sales',
    index='Category', 
    columns='Region',
    aggfunc = 'sum'
).round(2)
```

### Q4. Create a pivot table showing average profit by Segment and Category

```python
pd.pivot_table(
    df,
    values='Profit',
    index='Segment', 
    columns='Category',
    aggfunc = 'mean'
).round(2)
```

### Q5. Which month of the year has the highest total sales across all years?

```python
df.groupby('Order Month')['Sales'].sum().sort_values(ascending=False).head(1)
```

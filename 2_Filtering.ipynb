# Group2. Filtering with Multiple Conditions

### Q1. Find all orders from the 'West' region

```python
df[df['Region'] == 'West']
```

### Q2. Find all orders where Sales > 1000

```python
df[df['Sales'] > 1000]
```

### Q3. Find all orders where Category is 'Technology' AND Profit > 

```python
df[(df['Category'] == 'Technology') & (df['Profit'] > 0)]
```

### Q4. Find all orders where Region is 'East' OR 'West'

```python
df[df['Region'].isin(['East','West'])]
```

### Q5. Find all orders where Discount is NOT 0

```python
df[df['Discount'] == 0]
```

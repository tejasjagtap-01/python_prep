### Q1. Profitability by Segment + Region combined

```python
segment_region = df.groupby(['Segment','Region']).agg(
    total_profit = ('Profit','sum'),
    total_sales = ('Sales','sum'),
    order_count = ('Order ID','count')
).round(2)
segment_region
```

```python
segment_region['profit_margin%'] = (
    segment_region['total_profit'] / segment_region['total_sales'] * 100
).round(2)
print(segment_region.sort_values('profit_margin%', ascending=False))
```

### Q2. Correlation between Discount and Profit

```python
print(f"Discussed Profit Correlation : {df['Discount'].corr(df['Profit']).round(3)} ")
```

### Q3. Average profit margin for orders with discount > 0.3 vs discount = 0

```python
df["Profit Margin%"] = df['Profit'] / df['Sales'] * 100
df.head(2)
```

```python
df = df.drop(columns=['Profir Margin%'])
```

```python
df.head(1)
```

```python
high_discount = df[df['Discount'] > 0.3]['Profit Margin%'].mean()
no_discount = df[df['Discount'] == 0]['Profit Margin%'].mean()

print(f"Avg Margin with high discount: {high_discount:.2f}%")
print(f"Avg Margin with no discount: {no_discount:.2f}%")
```

### Q4. Flag orders as 'Healthy' (margin > 10%), 'Breakeven' (0-10%), or 'Loss' (< 0%)

```python
df['Health Flag'] = df['Profit Margin%'].apply(
    lambda x: 'Healthy' if x > 10 
        else ('Breakeven' if x >= 0 else 'Loss')
)
```

```python
df.head()
```

### Q5. Show count and total sales for each flag category

```python
df.groupby('Health Flag').agg(
    order_count = ('Order ID','count'), 
    total_sales = ('Sales','sum')
).round(2)
```

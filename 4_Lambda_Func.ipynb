### Q1. Create a new column 'Profit Margin %' = (Profit / Sales) × 100, rounded to 2 decimal places 

```python
df['Profit Margin %'] = (df['Profit'] / df['Sales'] * 100).round(2)
df.sample(2)
```

### Q2. Create a new column 'Performance' — 'Profitable' if Profit > 0, else 'Loss'

```python
df['Performance'] = df['Profit'].apply(
    lambda x : 'Profitable' if x > 0 else 'Loss'
)
```

```python
df.sample(3)
```

### Q3. Create a new column 'High Value Order' — True if Sales > 500, else False

```python
df['High Value Order'] = df['Sales'].apply(
    lambda x: 'High Value Sales' if x > 500 else ' False'
)
```
